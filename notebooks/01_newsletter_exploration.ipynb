# Newsletter Exploration

Explore the structure and content of SKUFood newsletter `.docx` files.
This notebook helps understand the data before building the parser.

```python
from docx import Document

from food_cpg_intelligence.config import settings
```

```python
# List available newsletter files
newsletters_dir = settings.resolve_path(settings.newsletters_dir)
docx_files = sorted(newsletters_dir.glob("*.docx"))
print(f"Found {len(docx_files)} newsletter files")
for f in docx_files[:5]:
    print(f"  {f.name} ({f.stat().st_size / 1024:.0f} KB)")
```

```python
# Load a sample newsletter and inspect structure
sample = docx_files[0]
doc = Document(sample)

print(f"File: {sample.name}")
print(f"Paragraphs: {len(doc.paragraphs)}")
print(f"Tables: {len(doc.tables)}")
print(f"Sections: {len(doc.sections)}")
print("\n--- Paragraph styles ---")
for i, p in enumerate(doc.paragraphs[:20]):
    print(f"  [{i:3d}] style={p.style.name!r:30s} text={p.text[:80]!r}")
```

```python
# Extract all text and measure size
full_text = "\n".join(p.text for p in doc.paragraphs if p.text.strip())
words = full_text.split()
print(f"Total words: {len(words)}")
print(f"Total chars: {len(full_text)}")
print("\n--- First 500 chars ---")
print(full_text[:500])
```

```python
# Check filename patterns for blog number extraction
import re

pattern = re.compile(r"[Ss][Kk][Uu][Ff]ood\s*blog\s*(\d+)", re.IGNORECASE)
for f in docx_files[:10]:
    match = pattern.search(f.stem)
    blog_num = match.group(1) if match else "NO MATCH"
    print(f"  {f.name} -> blog #{blog_num}")
```
